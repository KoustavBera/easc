```python
#3

# 1. Create temperature dataset (8 locations, 7 days)
temp = np.array([
    [30, 32, 34, 36, 33, 35, 37],
    [28, 30, 31, 33, 34, 32, 35],
    [36, 38, 37, 39, 35, 40, 38],
    [29, 31, 30, 32, 33, 31, 34],
    [34, 35, 36, 37, 38, 36, 39],
    [27, 29, 28, 30, 31, 29, 32],
    [32, 33, 35, 34, 36, 37, 35],
    [31, 30, 32, 33, 34, 32, 31]
])



#2. avg temp at each location
avg_temp = list(round(x) for x in (np.mean(temp, axis=1)))
print(f"average temp:{avg_temp}")

# 3. Maximum and minimum at each location
mx = np.max(temp, axis=1)
mn = np.min(temp, axis=1)
print(f"maximum temp:{mx}")
print(f"minimum temp:{mn}")

# 4. Location with highest average temperature
loc_max = np.argmax(avg_temp);
loc_max_temp = np.max(avg_temp)
print(f"Location {loc_max} has the highest recorded avg temp of {loc_max_temp} or {avg_temp[loc_max]}.")

# count temp > 35
count_gt_35 = np.sum(temp>35)
print(f"{count_gt_35} temperatures are >35:\n{temp[temp>35]}")

```

```python
#5
#1
A = np.array([
	[2, 1, 1],
	[1, 3, 2],
	[1, 4, 4],
])

B = np.array([10, 15, 20])

#2
det_A = np.linalg.det(A)
print(f"Determinant of {A} and {B} is {det_A}")

#3
if det_A != 0:
	print(f"Yes inv exists. Inverse(f(x)): {np.linalg.inv(A)}")
else:
	print("No inv DNE.")


#4
#solve the equation
X = np.linalg.solve(A, B);
print(f"Solution:{X}")

#Verify
print(f"AX={A @ X}")
print(f"B={B}")
```

```python
#6
A = np.array([2, 1, 3])
B = np.array([1, -1, 2])
C = np.array([3, 2, -1])
#1
A_dot_B = np.dot(A, B)
print(f"A.B = {A_dot_B}")

#2
A_x_B = np.cross(A, B)
print(f"A X B = {A_x_B}")

#3
# A.(BxC)
A_dot_B_x_C = np.dot(A, np.cross(B, C))
print(f"A.(BXC) = {A_dot_B_x_C}")

# (AXB).C
A_x_B_dot_C = np.dot(np.cross(A,B), C);
print(f"(AXB).C = {A_x_B_dot_C}")

print(f"A.(BXC)==(AXB).C is {A_x_B_dot_C==A_dot_B_x_C}")
```

```python
#7
F = np.array([10, 30, 20])
r = np.array([2, 3, 1])
d = np.array([3, 1, 2])

# torque = r x F
torq = np.cross(r, F);
magnitute = np.linalg.norm(torq)
print(f"Torq:{torq} and Magnitude: {round(magnitute)}")

# Work = F.d
work = np.dot(F, d)
print(f"Work: {work}")

# work = |F||d|cos(theta)
#cos(theta) = work/Fd
cos_theta = work/(np.linalg.norm(F)*np.linalg.norm(r))
theta = np.degrees(np.arccos(cos_theta))
print(f"Angle: {round(theta)} degrees")
```

```python
# 8 

import numpy as np

A = np.array([2, 1, 3])
B = np.array([1, -1, 2])

# Dot product: scalar
dot = np.dot(A, B)

# Cross product: vector
cross = np.cross(A, B)

# Dyadic product: matrix
dyadic = np.outer(A, B)

print("Dot product:", dot)
print("Cross product:", cross)
print("Dyadic product:\n", dyadic)
```

```python
#9

vd = np.array([12, 5])
vw = np.array([3, -2])

# va = actual velocity
va = vd + vw
print(f"Actual velocity: {va} m/s")

# magnitude
va_mag = np.linalg.norm(va)
print(f"Magnitude of actual velocity: {round(va_mag)} m/s")

# unit vector = va / mag

unit_vector = va / va_mag
print(f"Unit vector: {unit_vector} unit")

cos_theta = np.dot(va, vw)/(np.linalg.norm(va)*np.linalg.norm(vw))
theta = np.degrees(np.arccos(cos_theta))
print(f"Angle between actual velocity and wind velocity: {round(theta)} degrees")

```
