```python
# Q11

students = [
("S101", "Amit", [78, 85, 91, 67]),
("S102", "Neha", [92, 89, 95, 90]),
("S103", "Ravi", [56, 64, 71, 60]),
("S104", "Priya", [85, 90, 84, 88])
]

students_dict = {
    sid: {
        "name" : name,
        "marks" : marks,
    }
    for sid, name, marks in students 
}

print(f"After converting to dict: \n${students_dict}\n")

#• Calculate the total and average marks of every student and identify the student with the highest average


print(f"SID  Name  Total Marks Avg")
print("\n")
for sid in students_dict:
    marks = students_dict[sid]["marks"]
    total_marks = sum(marks)
    average_marks = total_marks / len(marks)
    students_dict[sid]["total"] = total_marks
    students_dict[sid]["average"] = average_marks
    print(sid, students_dict[sid]["name"], total_marks, average_marks)


# highest average.
highest = max(students_dict, key=lambda sid: students_dict[sid]["average"])
print(f"Highest average : {students_dict[highest]["average"]} acquired by {students_dict[highest]["name"]}")


# Using filter() and a lambda function, create a collection containing students whose
# average marks are greater than 80.
student_with_avg_g_80 = list(filter(lambda sid: students_dict[sid]["average"]>80, students_dict))
print(f"students whose marks are greater than 80: \n {student_with_avg_g_80}")

# • Using sorted() and a lambda function, sort the students according to their average
# marks
sorted_by_marks = sorted(students_dict, key=lambda sid: students_dict[sid]["marks"], reverse=True)
print(f"students sorted by marks:\n {sorted_by_marks} ")

# • Create a set containing the IDs of students who scored more than 85 in at least one
# subject.

student_with_marks_atleast_85 = set()

for sid in students_dict:
    for marks in students_dict[sid]["marks"]:
        if marks>85:
            student_with_marks_atleast_85.add(sid)
            break;


print(f"student with marks atleast >85 in one subject : \n {student_with_marks_atleast_85}")
```

```python
# 15

# Implement methods for deposit(), withdraw(), and displaying account details.
# Ensure
# that invalid transactions are rejected.

class BankAccount:
	def __init__(self, acc_no, holder, balance):
		self.acc_no = acc_no
		self.holder = holder

# • Encapsulate the balance so that it cannot be directly modified from outside the class
		self.__balance = balance

	@property
	def balance(self):
		return self.__balance

	def deposit(self, amount):
		if amount > 0:
			self.__balance += amount
			print("Deposit successful")
		else:
			print("Invalid deposit amount")
			
	def withdraw(self, amount):
		if amount > 0 and amount <= self.__balance:
			self.__balance -= amount
			print("Withdrawal successful")
		else:
			print("Invalid withdrawal")

	def display(self):
		print(f"{self.acc_no}--{self.holder}--{self.balance}")

	def add_interest(self):
		pass

# • Create a subclass SavingsAccount with an interest rate and override an appropriate
# method to calculate/add interest.
class SavingsAccount(BankAccount):
	def __init__(self, acc_no, holder, balance, rate):
		super().__init__(acc_no, holder, balance)
		self.rate = rate

	def add_interest(self):
		interest = self.balance * self.rate / 100
		self.deposit(interest)
		print(f"Interest of {interest} added to the account.")

#  Create multiple account objects 
# 
accounts = [
	BankAccount("A101", "Amit", 5000),
	SavingsAccount("A102", "Neha", 10000, 5)
]

# Deposit and withdrawal
accounts[0].deposit(1000)
accounts[1].withdraw(500)

#add interest to savings account
for acc in accounts:
	if isinstance(acc, SavingsAccount):
		acc.add_interest()

#Display account
print("\n Account Details:")
for acc in accounts:
	acc.display()



# and store them in a list of objects. 
# 
# Use a lambda
# function with sorted() to sort accounts according to their balance.

sorted_accounts = sorted(accounts, key=lambda acc: acc.balance, reverse=True)

print("\nAccounts sorted by balance:")
for acc in sorted_accounts:
	acc.display()




# • Create a dictionary mapping account types to the set of account numbers belonging
# to each type.

#Dictionary mapping account types to sets of account number
account_types = {"Regular": set(), "Savings": set()}

for acc in accounts:
	if isinstance(acc, SavingsAccount):
		account_types["Savings"].add(acc.acc_no)

	else:
		account_types["Regular"].add(acc.acc_no)

print(f"\nAccount types: ", account_types)



```
