```python
#1

# Generate x values
x = np.linspace(-5, 5, 100)

# Calculate the functions
y1 = x**2
y2 = 3*x

# Create Figure and Axes
fig, ax = plt.subplots()

# Plot both curves
ax.plot(x, y1, color='red', linewidth=3, label='y1 = x^2')
ax.plot(x, y2, color='green', linewidth=2, label='y2 = 3x')

# Add labels, title and legend
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Plot of y1 = x^2 and y2 = 3x')
ax.legend()

# Display the graph
plt.show()
```

```python
#8
""" 
Purpose of add_subplot():
The add_subplot() method is used to add an axes(drawing area) to an existing figure at a specified position in a grid.
Example : fig.add_suplot(2,2,1) adds an axes in the 1st cell of a 2x2 grid
"""
x = np.linspace(5,10,100)
y1 = x
y2 = x ** 2
y3 = x ** 3

fig = plt.figure()
ax1 = fig.add_subplot(3,1,1)
ax1.plot(x,y1)
ax1.set_title("y = x")

ax2 = fig.add_subplot(3,1,2)
ax2.plot(x, y2)
ax2.set_title("y = x^2")

ax2 = fig.add_subplot(3,1,3)
ax2.plot(x, y3)
ax2.set_title("y = x^3")


plt.tight_layout()
plt.show()
```

```python

#9
""" 
Consider the following Python statements:
fig, ax = plt.subplots(2, 2)
1. How many Axes objects are created?
->4 axes objects are created because 2rows 2 cols= 2x2 4 objects

2. What is the shape of the ax object?
-> The shape of the ax object is (2, 2) because it is a 2D array of Axes objects with 2 rows and 2 columns.

3. How would you access the Axes object in the first row and second column?
-> You can access the Axes object in the first row and second column using ax[0, 1].

4. How would you access the Axes object in the second row and first column?
-> You can access the Axes object in the second row and first column using ax[1, 0].

5.Write statements to plot (y=x**2
) in the first subplot and (y=x**3
) in the last subplot.
"""
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2)

x = np.linspace(0, 5, 100)

# First subplot: y = x^2
ax[0, 0].plot(x, x**2)
ax[0, 0].set_title('y = x^2')

# Last subplot: y = x^3
ax[1, 1].plot(x, x**3)
ax[1, 1].set_title('y = x^3')

plt.tight_layout()
plt.show()
```

```python
#10

import numpy as np
import matplotlib.pyplot as plt

# Create four Axes objects in a 2 x 2 arrangement
fig, ax = plt.subplots(2, 2)

# Generate x values
x = np.linspace(0, 5, 100)

ax[0,0].plot(x, x**2, color="red")
ax[0,0].set_xlabel("x")
ax[0,0].set_ylabel("y")
ax[0,0].set_title("y = x^2")
ax[0,0].grid(True)


ax[0,1].plot(x, 2*x + 1, color="blue")
ax[0,1].set_xlabel("x")
ax[0,1].set_ylabel("y")
ax[0,1].set_title("y = 2x + 1")
ax[0,1].grid(True)


ax[1,0].plot(x, np.sin(x), color="green")
ax[1,0].set_xlabel("x")
ax[1,0].set_ylabel("y")
ax[1,0].set_title("y = sin(x)")
ax[1,0].grid(True)


ax[1,1].plot(x, np.cos(x), color="black")
ax[1,1].set_xlabel("x")
ax[1,1].set_ylabel("y")
ax[1,1].set_title("y = cos(x)")
ax[1,1].grid(True)

plt.tight_layout()
plt.show()
```

```python
#11
"""  
Differentiate between the subplots() function and the add_subplot() method in Matplotlib
-> 
subplots()
1. Creates a figure and multiple axes together
2. Creates Fig object automatically
3. access axes by indexing eg., ax[0,0]
4. creates all specified subplots at once
ex: fig, ax = plt.subplots(2,2)

add_subplots()
1. adds an axes to a existing fig object
2. Requires fig object to be created first
3. each add_subplot() call returns a object with its position
4. create subplot one by one
ex: ax1 = fig.add_subplot(row, column, position)

"""

# Method 1. plt.subplots()

import matplotlib.pyplot as plt

# Create a Figure with four Axes
fig, ax = plt.subplots(2, 2)

# Add titles to each subplot
ax[0, 0].set_title('First Subplot')
ax[0, 1].set_title('Second Subplot')
ax[1, 0].set_title('Third Subplot')
ax[1, 1].set_title('Fourth Subplot')

plt.tight_layout()
plt.show()


# Method 2. fig.add_subplot()


import matplotlib.pyplot as plt

# Create a Figure
fig = plt.figure()

# Add four Axes individually
ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)

# Add titles to each subplot
ax1.set_title('First Subplot')
ax2.set_title('Second Subplot')
ax3.set_title('Third Subplot')
ax4.set_title('Fourth Subplot')

plt.tight_layout()
plt.show()
```

```python
#12

fig = plt.figure()

ax1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])

x = np.linspace(-5, 5, 100)
y = np.sin(x)

ax1.plot(x, y)
ax1.set_title("Main plot")
ax1.set_xlabel("x")
ax1.set_ylabel("y")

ax2 = ax1.inset_axes([0.5, 0.5, 0.4, 0.4])

ax2.plot(x, y)
ax2.set_xlim(2, 4)
ax2.set_xlabel("x")
ax2.set_ylabel("y")
ax2.set_title("smaller one")
ax2.grid(True)

plt.show()
```

```python
#15
fig, ax = plt.subplots()

x = np.linspace(0,10,500)

y1 = (x - 3)**2
y2 = -(x - 2)**2 + 8

ax.plot(x, y1, label="y1", color="red")
ax.plot(x, y2, label="y2", color="blue")
ax.set_xlabel("x")
ax.set_ylabel("y")

ax.fill_between(x, y1, y2, where=(y2>=y1), color="yellow", alpha=0.5)
ax.grid(True)
ax.legend()
plt.show()
```
